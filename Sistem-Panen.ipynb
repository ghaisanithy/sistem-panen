```python
def hitung_pendapatan_kentang(berat_kg, harga_per_kg):
    return berat_kg * harga_per_kg

if __name__ == "__main__":
    komoditas = "Kentang"
    berat_panen = 1200     # berat dalam kilogram (kg)
    harga_per_kg = 14000   # harga jual per kg (Rp)

    total_pendapatan = hitung_pendapatan_kentang(berat_panen, harga_per_kg)

    print("=" * 35)
    print(f"Laporan Pendapatan Panen {komoditas}")
    print("=" * 35)
    print(f"Komoditas        : {komoditas}")
    print(f"Total Panen      : {berat_panen:,} kg")
    print(f"Harga Jual       : Rp {harga_per_kg:,} / kg")
    print("-" * 35)
    print(f"Total Pendapatan : Rp {total_pendapatan:,}")
    print("=" * 35)
```
